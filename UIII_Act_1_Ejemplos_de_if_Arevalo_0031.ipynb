## Ejemplos de if

```python
numero = int(input("Ingresa un número: ")) # Convierte la entrada del usuario a entero [12].

if numero > 0:                             # Evalúa si el número es mayor que 0 (relacional) [12].
    print("El número es positivo.")        # Se ejecuta si la condición anterior es verdadera [12].
elif numero < 0:                           # Evalúa si es menor que 0 si la primera fue falsa [12].
    print("El número es negativo.")        # Se ejecuta si el número es menor a 0 [12].
else:                                      # Se ejecuta si ninguna de las condiciones anteriores se cumplió [12].
    print("El número es cero.")            # Resultado cuando el número es exactamente 0 [12].
```

```python
año = int(input("Ingresa un año: ")) # Solicita el año al usuario [13].

# Verifica si es divisible por 4 y no por 100, O si es divisible por 400 (aritmético y lógico) [13].
if (año % 4 == 0 and año % 100 != 0) or (año % 400 == 0):
    print(f"{año} es un año bisiesto.") # Se imprime si la lógica combinada resulta verdadera [13].
else:                                   # Rama para cuando la condición lógica es falsa [13].
    print(f"{año} no es un año bisiesto.") # Resultado para años comunes [13].
```

```python
cantidad = int(input("¿Cuántos artículos compraste? ")) # Lee la cantidad de artículos [12].

if cantidad >= 10:               # Verifica si la cantidad es mayor o igual a 10 [12].
    descuento = 0.2              # Asigna 20% de descuento si se cumple [12].
elif cantidad >= 5:              # Verifica si es mayor o igual a 5 (pero menor a 10) [12].
    descuento = 0.1              # Asigna 10% de descuento [12].
else:                            # Para cualquier cantidad menor a 5 [12].
    descuento = 0                # Sin descuento [12].

print(f"Tu descuento es del {descuento * 100}%.") # Muestra el resultado final calculado [13].
```

```python
numero = int(input("Ingresa un número: ")) # Captura el número [14].

if 10 <= numero <= 20:                     # Verifica si el número está entre 10 y 20 inclusive [14].
    print(f"El número {numero} está entre 10 y 20.") # Ejecución si está en el rango [14].
else:                                      # Rama si el número está fuera del intervalo [14].
    print(f"El número {numero} no está entre 10 y 20.") # Ejecución si está fuera [14].
```
